# MORROW — Exploratory Data Analysis

This notebook analyzes the cleaned synthetic hospital appointment dataset (`data/processed/synthetic_appointments_cleaned.csv`).
All data is strictly synthetic (`random_seed = 42`).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/processed/synthetic_appointments_cleaned.csv', parse_dates=['booking_created_at', 'appointment_datetime'])
print(f'Total cleaned records: {len(df)}')
df.info()
```

```python
df.describe().T
```

```python
print('Target Distribution:')
print(df['no_show'].value_counts(normalize=True))

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df['no_show'].value_counts().plot(kind='bar', ax=axes[0], color=['#8D9E8F', '#A85D48'])
axes[0].set_title('Attendance vs No-Show Count')

df.groupby('booking_channel')['no_show'].mean().plot(kind='bar', ax=axes[1], color='#1F5A57')
axes[1].set_title('No-Show Rate by Booking Channel')
plt.tight_layout()
plt.show()
```

```python
num_cols = [
    'booking_lead_days', 'appointment_weekday', 'appointment_hour',
    'booking_hour', 'prior_appointments_count', 'prior_no_show_count',
    'prior_cancellations_count', 'prior_no_show_rate', 'days_since_last_appointment', 'no_show'
]
plt.figure(figsize=(10, 8))
sns.heatmap(df[num_cols].corr(), annot=True, fmt='.2f', cmap='RdBu_r', center=0)
plt.title('Feature Correlation Heatmap')
plt.show()
```
